```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://quotes.toscrape.com"

response= requests.get(url)
```

```python
print(f"Status:{response.status_code}")
print(f"Status:{response.encoding}")
```

```python
soup = BeautifulSoup(response.text,'html.parser')

data=[]
citations = soup.find_all("div",class_='quote')

for citation in citations :
    citation_texte = citation.find('span',class_='text').text
    auteur = citation.find('small',class_='author').text
    tag_links = citation.find_all('a',class_='tag')
    tags = []
    for tag in tag_links:
        tags.append(tag.text)
    data.append({'citation':citation_texte,'auteur':auteur,'tags':",".join(tags)})
```

```python
data[0]
```

```python
import pandas  as pd
data = pd.DataFrame(data=data)
data.head()
```

```python
data.to_csv("./quotes.csv",index=False)
```
