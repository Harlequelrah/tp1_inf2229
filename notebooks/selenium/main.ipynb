```python
from selenium.webdriver.common.by import By
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
options = webdriver.ChromeOptions()
options.add_argument("--headless") # sans interface graphique
options.add_argument("--no-sandbox")
```

```python
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)
```

```python
try :
    driver.get("https://quotes.toscrape.com/scroll") #Page avec
    citations = set()
    dernier_count=0
    tentatives = 0

    while tentatives < 5 :
        driver.execute_script("window.scrollTo(0,document.body.scrollHeight);")
        time.sleep(2)

        elements = driver.find_elements(By.CLASS_NAME,"quote")

        for el in elements :
            texte = el.find_element(By.CLASS_NAME,"text").text
            auteur = el.find_element(By.CLASS_NAME,"author").text
            citations.add((texte,auteur))

        if len(citations) == dernier_count:
            tentatives+=1
        else:
            dernier_count = len(citations)
            tentatives = 0
    for texte,auteur in list(citations)[:10]:
        print(f"{texte} \n - {auteur} \n")
finally :
    driver.quit()
```
