```python
import requests
from bs4 import BeautifulSoup
import csv

BASE_URL = "https://books.toscrape.com/catalogue/"
les_livres = []
page = 1

while True:
    url = f"{BASE_URL}page-{page}.html"
    r = requests.get(url)
    if r.status_code != 200:
        print(f"(Fin de la page {page - 1})")
        break

    soup = BeautifulSoup(r.text, "html.parser")
    livres = soup.find_all("article", class_="product_pod")

    for livre in livres:
        titre = livre.h3.a["title"]
        prix = livre.find("p", class_="price_color").text
        dispo = livre.find("p", class_="availability").text.strip()
        note = livre.p["class"][1]

        les_livres.append(
            {"titre": titre, "prix": prix, "disponibilité": dispo, "note": note}
        )

    suivant = soup.find("li", class_="next")
    if suivant is None:
        break

    page += 1
    print(f"Page {page} récupérée...")

with open("livres.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["titre", "prix", "disponibilité", "note"])
    writer.writeheader()
    writer.writerows(les_livres)

print(f"{len(les_livres)} livres sauvegardés dans le fichier csv.")
```
