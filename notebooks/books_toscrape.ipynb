```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://books.toscrape.com/"

response= requests.get(url)
```

```python
print(f"Status:{response.status_code}")
print(f"Status:{response.encoding}")
```

```python
soup = BeautifulSoup(response.text,'html.parser')

livres = soup.find_all("article",class_='product_pod')

for livre in livres :
    titre =  livre.h3.a["title"]
    prix = livre.find("p",class_="price_color").text
    dispo = livre.find("p", class_="instock availability").text.strip()
    print(f"titre | prix | dispo")
    print(f"{titre} | {prix} | {dispo}")
```
